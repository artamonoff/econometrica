## Спецификация модели

Линейная спецификация с бинарной зависимой переменной `y`

$$
	y=x^\top\beta+u=\beta_0+\beta_1x_1+\cdots+\beta_kx_k+u
$$

Альтернативная спецификация

$$
	P(y=1)=x^\top\beta=\beta_0+\beta_1x_1+\cdots+\beta_kx_k
$$

```python

import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

from statsmodels.iolib.summary2 import summary_params, summary_col # вывод результатов тестирования

# критические значения
from scipy.stats import t, f

# Не показывать Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Не показывать ValueWarning, ConvergenceWarning из statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)

from IPython.display import Markdown, display
```

## Подгонка модели

```python
df_names = ['loanapp', 'loanapp', 'TableF5-1', 'TableF5-1']

endog_var = ['approve', 'approve', 'LFP', 'LFP']

exog_vars =['mortno+unem+dep+male+married+yjob+self', 
			'appinc+I(appinc**2)+mortno+unem+dep+male+married+yjob+self',
			'WA+I(WA**2)+WE+KL6+K618+CIT+UN+np.log(FAMINC)',
			'WA+WE+CIT+UN+np.log(FAMINC)']

mod_name = ['Approve equation №1', 'Approve equation №2', 'labour force equation №1',  'labour force equation №2']

digits = 3

for i, df_name in enumerate(df_names):
	df = pd.read_csv('../datasets/'+df_name+'.csv')

	mod = ols(formula=endog_var[i]+'~'+exog_vars[i], data=df)
	res = mod.fit()

	display(Markdown(f'### {mod_name[i]}'))
	display(Markdown(f'''Для датасет `{df_name}` оцените __LPM__ регрессию `{endog_var[i]}` на `{exog_vars[i]}`
	и укажите её коэффициенты. __Ответ округлите до {digits} десятичных знаков__ '''))

	df_coefs = pd.DataFrame({'Коэффициент':res.params.index, 'Оценка':res.params.round(digits)})
	display(Markdown(f'''Ответ:\n'''))
	display(Markdown(df_coefs.to_markdown(index=False)))
	# print(df_coefs.to_markdown(index=False))
	display(Markdown(f'''__Дайте интерпретацию коэффициентам модели__'''))
```

## t-тест

```python
df_names = ['loanapp', 'loanapp', 'TableF5-1', 'TableF5-1']

endog_var = ['approve', 'approve', 'LFP', 'LFP']

exog_vars =['mortno+unem+dep+male+married+yjob+self', 
			'appinc+I(appinc**2)+mortno+unem+dep+male+married+yjob+self',
			'WA+I(WA**2)+WE+KL6+K618+CIT+UN+np.log(FAMINC)',
			'WA+WE+CIT+UN+np.log(FAMINC)']

mod_name = ['Approve equation №1', 'Approve equation №2', 'labour force equation №1',  'labour force equation №2']

digits = 3

sign_level = [0.01, 0.01, 0.01, 0.05]

for i, df_name in enumerate(df_names):
	df = pd.read_csv('../datasets/'+df_name+'.csv')

	mod = ols(formula=endog_var[i]+'~'+exog_vars[i], data=df)
	res = mod.fit(cov_type='HC3')

	display(Markdown(f'### {mod_name[i]}'))
	display(Markdown(f'''Для датасет `{df_name}` оцените __LPM__ регрессию `{endog_var[i]}` на `{exog_vars[i]}`
	и тестируйте значимость её коэффициентов. Уровень значимости {sign_level[i]*100}%, число наблюдений {int(mod.nobs)}. 
	Для каждого коэффициента укажите его тестовую статистику, P-значение
	и седлайте вывод. Найдите соответствующее критическое значение.  __Ответ округлите до {digits} десятичных знаков__ '''))

	res_t_test = pd.DataFrame({'Коэффициент':res.params.index, 
								'Тест стат':res.tvalues.round(digits),
								'P-значение':res.pvalues.round(digits),
								'Вывод': res.pvalues.apply(lambda x: 'Значим' if x<sign_level[i] else 'Незначим') })
	display(Markdown(f'''Ответ:\n'''))
	display(Markdown(res_t_test.to_markdown(index=False)))
	# print(df_coefs.to_markdown(index=False))

	crit_val = t.isf(q=sign_level[i]/2, df=res.df_resid).round(digits)
	display(Markdown(f'''Критическое значение: {crit_val}'''))
```

## F-тест: Значимость регрессии

Для регрессии

$$
	P(y=1)=x^\top\beta=\beta_0+\beta_1x_1+\cdots+\beta_kx_k
$$

тестируем гипотезу

$$
	H_0:\beta_1=\cdots=\beta_k=0
$$

```python
df_names = ['loanapp', 'loanapp', 'loanapp']

endog_var = ['approve', 'approve', 'approve']

exog_vars =['unem+male+yjob+self', 
			'appinc+I(appinc**2)+mortno+dep',
			'dep+male+married']

mod_name = ['Approve equation №1', 'Approve equation №2', 'Approve equation №3']

digits = 3

sign_level = [0.10, 0.05, 0.01]

for i, df_name in enumerate(df_names):
	df = pd.read_csv('../datasets/'+df_name+'.csv')

	mod = ols(formula=endog_var[i]+'~'+exog_vars[i], data=df)
	res = mod.fit(cov_type='HC3')

	display(Markdown(f'### {mod_name[i]}'))
	display(Markdown(f'''Для датасет `{df_name}` оцените __LPM__ регрессию `{endog_var[i]}` на `{exog_vars[i]}`.
	Тестируйте значимость регрессии. Уровень значимости {sign_level[i]*100}%. 
	Найдите значение тестовой статистики, её P-значение, критическое значение и сделайте вывод.
	__Ответ округлите до {digits} десятичных знаков__ '''))

	test_stat = res.fvalue
	p_val = res.f_pvalue
	crit_val = f.isf(sign_level[i], dfn=res.df_model, dfd=res.df_resid)
	is_significant = 'значима' if p_val<sign_level[i] else 'незначима' 

	res_f_test_overall = pd.DataFrame({'Статистика': [round(test_stat, digits)], 
										'P-значение': [round(test_stat, digits)],
										'Критическое': [crit_val.round(digits)], 
										'Регрессия': [is_significant]})

	display(Markdown(f'''Ответ:\n'''))
	display(Markdown(res_f_test_overall.to_markdown(index=False)))
	# display(Markdown(f'Тестовая статистика = {round(test_stat, digits)}'))
	# display(Markdown(f'P-значение = {round(p_val, digits)}'))
	# display(Markdown(f'Критическое значение = {crit_val.round(digits)}'))
	# display(Markdown(f'Вывод: регрессия __{is_significant}__'))
	# print(df_coefs.to_markdown(index=False))
```

## F-тест: Совместная значимость

```python
df_names = ['loanapp']

endog_var = ['approve']

exog_vars =['appinc+I(appinc**2)+mortno+unem+dep+male+married+yjob+self']

mod_name = ['Approve equation №1']

digits = 3

sign_level = [[0.05, 0.01]]

joint_sign = [ ['appinc, appinc**2', 'unem, dep, married'] ]
hypotheses = [ ['appinc=0, I(appinc ** 2)=0', 'unem=0, dep=0, married=0'] ]

for i, df_name in enumerate(df_names):
	df = pd.read_csv('../datasets/'+df_name+'.csv')

	mod = ols(formula=endog_var[i]+'~'+exog_vars[i], data=df)
	res = mod.fit()

	display(Markdown(f'### {mod_name[i]}'))
	display(Markdown(f'''Для датасет `{df_name}` оцените __LPM__ регрессию `{endog_var[i]}` на `{exog_vars[i]}`
	Тестируйте гипотезы ниже. Для каждой гипотезы дайте интерпретацию, найдите тестовую статистику, её P-значение, критическое значение и сделайте вывод.
	 __Ответ округлите до {digits} десятичных знаков__ '''))

	for j, hypothesis in enumerate(hypotheses[i]):
		display(Markdown(f'#### Гипотеза №{j+1}'))
		display(Markdown(f'Тестируйте совместную значимость `{joint_sign[i][j]}`'))

		res_f_test = res.wald_test(hypothesis, use_f=True)

		test_stat = res_f_test.statistic
		p_val = res_f_test.pvalue
		crit_val = f.isf(sign_level[i][j], dfn = res_f_test.df_denom, dfd = res.df_resid)
		is_significant = 'значимо' if p_val<sign_level[i][j] else 'незначимо'

		res_f_test_joint = pd.DataFrame({'Статистика': [round(test_stat, digits)], 
										'P-значение': [round(test_stat, digits)],
										'Критическое': [crit_val.round(digits)], 
										'Совместное влияние': [is_significant]})

		display(Markdown(f'''Ответ:\n'''))
		display(Markdown(res_f_test_joint.to_markdown(index=False)))
```
