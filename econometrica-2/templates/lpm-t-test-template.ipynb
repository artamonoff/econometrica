# LPM-регрессия: t-тест

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols
from statsmodels.iolib.summary2 import summary_params # вывод результатов тестирования

from scipy.stats import t # критические значения t-распределения

# Не показывать Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Не показывать ValueWarning, ConvergenceWarning из statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# импорт данных
df = pd.read_csv('Filename.csv')
```

## Спецификация и подгонка

__Замечание__: при подгонке учитываем попреку на гетероскедастичность (используем робастную оценку ковариационной матрицы)

```python
# спецификация модели через формулу
mod_spec = 'your specification'
# инициализация модели
mod = ols(formula = mod_spec, data = df)
# подгонка модели под данные/обучение на данных
res = mod.fit(cov_type='HC3')
```

## t-тест

вывод результатов t-теста

```python
print( summary_params(res, alpha=0.05) )
```

результаты t-теста с округлением

```python
digits = 3
print( summary_params(res, alpha=0.05).round(digits) )
```

критическое значение t-распределения со степенями свободы $df=n-k-1$

```python
# уровень значимости
sign_level = 0.05
# Критическое значение t-распределения
crit_val = t.isf(q=sign_level/2, df=res.df_resid)
crit_val
```

```python
# критическое значение с округлением
crit_val.round(2)
```
