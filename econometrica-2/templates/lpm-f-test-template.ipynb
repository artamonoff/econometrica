# LPM-регрессия: F-тест

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

from scipy.stats import f # критические значения F-распределения

# Не показывать Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Не показывать ValueWarning, ConvergenceWarning из statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# импорт данных
df = pd.read_csv('Filename.csv')
```

## Спецификация и подгонка

__Замечание__: при подгонке учитываем попреку на гетероскедастичность (используем робастную оценку ковариационной матрицы)

```python
# спецификация модели через формулу
mod_spec = 'your specification'
# инициализация модели
mod = ols(formula = mod_spec, data = df)
# подгонка модели под данные/обучение на данных
res = mod.fit(cov_type='HC3')
```

## F-тест: значимость регрессии

F-статистика и её P-значение

```python
res.fvalue, res.f_pvalue
```

F-статистика и её P-значение с округлением

```python
digits = 3
round(res.fvalue, digits), round(res.f_pvalue, digits)
```

критическое значение F-распределения по степенями свобода $df1=k$, $df2=n-k-1$

```python
# уровень значимости
sign_level = 0.05
# Критическое значение F-распределения
crit_val = f.isf(sign_level, dfn=res.df_model, dfd=res.df_resid)
crit_val
```

```python
# критическое значение с округлением
crit_val.round(2)
```

## F-тест: совместная значимость

```python
F_test = res.wald_test('your hypothesis', use_f=True)
print(F_test)
```

Тестовая статистика и её P-значение

```python
F_test.statistic, F_test.pvalue
```

Тестовая статистика и её P-значение с округлением

```python
F_test.statistic.round(3), F_test.pvalue.round(3)
```

критическое значение F-распределения

```python
# уровень значимости
sign_level = 0.05
# Критическое значение F-распределения
crit_val = f.isf(sign_level, dfn = F_test.df_denom, dfd = res.df_resid)
```

```python
# критическое значение с округлением
crit_val.round(2)
```
