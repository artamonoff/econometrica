# logit/probit/LPM regression: prediction

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

# Do not show Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# data import
df = pd.read_csv('Filename.csv')
```

## Specification and fitting

```python
# model specification via formula
mod_spec = 'your specification'
# model initialization
mod = ols(formula = mod_spec, data = df)
# fitting the model to the data / training on the data
res = mod.fit(cov_type='HC3')
```

## Prediction

```python
# Create a dataframe with new regressor data for prediction
new_data = pd.DataFrame({'exog1': [ ], 'exog2': [ ] })
new_data
```

```python
# Prediction
res.predict(exog=new_data, transform=True)
```

```python
# Prediction, rounded
res.predict(exog=new_data, transform=True).round(3)
```
