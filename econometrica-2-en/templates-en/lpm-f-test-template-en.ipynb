# LPM regression: F-test

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

from scipy.stats import f # critical values of the F-distribution

# Do not show Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# data import
df = pd.read_csv('Filename.csv')
```

## Specification and fitting

__Note__: when fitting, we account for heteroskedasticity (use a robust estimate of the covariance matrix)

```python
# model specification via formula
mod_spec = 'your specification'
# model initialization
mod = ols(formula = mod_spec, data = df)
# fitting the model to the data / training on the data
res = mod.fit(cov_type='HC3')
```

## F-test: overall significance of regression

F-statistic and its P-value

```python
res.fvalue, res.f_pvalue
```

F-statistic and its P-value, rounded

```python
digits = 3
round(res.fvalue, digits), round(res.f_pvalue, digits)
```

critical value of the F-distribution with degrees of freedom $df1=k$, $df2=n-k-1$

```python
# significance level
sign_level = 0.05
# Critical value of the F-distribution
crit_val = f.isf(sign_level, dfn=res.df_model, dfd=res.df_resid)
crit_val
```

```python
# critical value, rounded
crit_val.round(2)
```

## F-test: joint significance

```python
F_test = res.wald_test('your hypothesis', use_f=True)
print(F_test)
```

Test statistic and its P-value

```python
F_test.statistic, F_test.pvalue
```

Test statistic and its P-value, rounded

```python
F_test.statistic.round(3), F_test.pvalue.round(3)
```

critical value of the F-distribution

```python
# significance level
sign_level = 0.05
# Critical value of the F-distribution
crit_val = f.isf(sign_level, dfn = F_test.df_denom, dfd = res.df_resid)
```

```python
# critical value, rounded
crit_val.round(2)
```
