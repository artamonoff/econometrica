# logit/probit regression: marginal effects

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import logit, probit

# Do not show FutureWarnings
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# import data
df = pd.read_csv('Filename.csv')
```

## Specification and fitting

```python
# model specification via formula
mod_spec = 'your specification'
# model type: logit or probit
mod_type = 'logit' # 'probit'
# initialize the model
if mod_type == 'probit':
	mod = probit(formula = mod_spec, data = df)
else:
	mod = logit(formula = mod_spec, data = df)
# fit the model to the data / train on the data
res = mod.fit(disp=False)
```

## Marginal effects
To compute them, the `at` parameter must be specified:
* ` 'mean' ` to compute the marginal effect at the mean point
* ` 'overall' ` to compute the average marginal effect over the sample

```python
margeff_res = res.get_margeff(at=, dummy=True)
# output the results
margeff_res.summary()
```

```python
# short report with rounding
margeff_res.summary_frame().round(3)
```
