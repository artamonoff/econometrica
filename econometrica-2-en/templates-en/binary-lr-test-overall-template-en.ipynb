# logit/probit regression: LR test, overall significance of regression

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import logit, probit

from scipy.stats import chi2 # critical values of the chi2 distribution

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# import data
df = pd.read_csv('Filename.csv')
```

## Specification and fitting

```python
# model specification via formula
mod_spec = 'your specification'
# model type: logit or probit
mod_type = 'logit' # 'probit'
# initialize the model
if mod_type == 'probit':
	mod = probit(formula = mod_spec, data = df)
else:
	mod = logit(formula = mod_spec, data = df)
# fit the model to the data / train on the data
res = mod.fit(disp=False)
```

## LR test: overall significance of regression
We test the hypothesis

$$ H_0: \beta_1=\cdots=\beta_k=0$$

When fitting the model, the LR statistic and its P-value are computed by default

```python
res.llr, res.llr_pvalue
```

LR statistic and its P-value with rounding

```python
digits = 3
round(res.llr, digits), round(res.llr_pvalue, digits)
```

critical value of the $\chi^2_{df}$ distribution

```python
# significance level
sign_level = 0.05
# Critical value of the chi-square distribution
crit_val = chi2.isf(q=sign_level, df=res.df_model)
crit_val
```

```python
# critical value with rounding
crit_val.round(2)
```

## Computational details
$LR=2(\log L-\log L_{null})$ where
- $\log L$ – log-likelihood for the original regression
- $\log L_{null}$ – log-likelihood for the regression on the constant only

Output $\log L, \log L_{null}, 2(\log L-\log L_{null})$

```python
res.llf, res.llnull, 2*(res.llf-res.llnull)
```
