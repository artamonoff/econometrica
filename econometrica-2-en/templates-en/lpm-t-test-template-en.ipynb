# LPM regression: t-test

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols
from statsmodels.iolib.summary2 import summary_params # output of test results

from scipy.stats import t # critical values of the t-distribution

# Do not show Warning
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# data import
df = pd.read_csv('Filename.csv')
```

## Specification and fitting

__Note__: when fitting, we account for heteroskedasticity (use a robust estimate of the covariance matrix)

```python
# model specification via formula
mod_spec = 'your specification'
# model initialization
mod = ols(formula = mod_spec, data = df)
# fitting the model to the data / training on the data
res = mod.fit(cov_type='HC3')
```

## t-test

output of t-test results

```python
print( summary_params(res, alpha=0.05) )
```

t-test results, rounded

```python
digits = 3
print( summary_params(res, alpha=0.05).round(digits) )
```

critical value of the t-distribution with degrees of freedom $df=n-k-1$

```python
# significance level
sign_level = 0.05
# Critical value of the t-distribution
crit_val = t.isf(q=sign_level/2, df=res.df_resid)
crit_val
```

```python
# critical value, rounded
crit_val.round(2)
```
