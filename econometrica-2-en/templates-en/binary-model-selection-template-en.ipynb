# logit/probit regression: Comparing models

```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import logit, probit
from statsmodels.iolib.summary2 import summary_col # output results of several regressions


# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

```python
# import data
df = pd.read_csv('Filename.csv')
```

## Specification and fitting of several models

```python
mod_specs = ['your specification 1', 'your specification 2', ]
mod_type = ['logit', 'probit', ]

# list for fitted models
res = []

for i, mod_spec in enumerate(mod_specs):
	if mod_type[i] == 'probit':
		mod = probit (formula = mod_spec, data = df)
	else:
		mod = logit (formula = mod_spec, data = df)

	res += mod.fit(disp=False)
```

## Output several models in one table including AIC, BIC, pseudo-R^2_adj

```python
# Model names
mod_names = ['Model 1', 'Model 2', ]
# order of regressors in the table
# reg_order = ['Intercept', ]
# dictionary with what to add
info_to_add = {'AIC': lambda x: '%.3f' % x.aic, 'BIC': lambda x: '%.3f' % x.bic, 'R2_adj': lambda x: '%.3f' % (1-(x.llf-x.df_model-1)/x.llnull) }
# the '%.3f' % parameter is responsible for rounding the number after the percent sign to 3 decimal places

summary_col(results=res, model_names=mod_names, stars=True, float_format='%.3f', # regressor_order=reg_order,
             info_dict=info_to_add)
```
